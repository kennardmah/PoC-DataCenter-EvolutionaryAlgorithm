### Intro

Dynamic Train Dispatching Problem (DTDP) can be seen as a Constrained Markov Decision Process (CMDP).

### Design

**Infrastructure State:**

For each $e \in E$:
- occupied $\in \{0, 1\}$
- train_id
- t_release (earlist free time)
- direction

for stations:
- platform occupancy vector
- platform compatibility mask
- dwell constraints

**Train State:**

{
    current_node: station / junction,
    current_edge: if in transit,
    next_possible_edges: adjacency mask,
    remaining_travel_time: if on edge,
    delay: vs timetable,
    priority_class: express/local,
    scheduled_arrival: target time
}

**Global Time Encoding**

- current simulation time $t$
- event queue snapshot (next $k$ events)

**Graph Encoding (for NN)**

- Graph Neural Network (GNN) for infrastructure
- Train embeddings pooled via attention

### Action Space

Decision Trigger:
- train reaches node
- block is released
- platform is free
- disturbance occurs

Action types:
- choose next edge $\in$ feasible outgoing edges
- holding control: wait $\delta t \in \{0, h, 2h\}$
- platform assignment: select platform $\in$ compatible platforms
- depart now | delay $\delta t$

Transition Function:
- edge traversal time is sampled from distribution
- dwell time sampled within bounds
- disturbances:
    - random delays
    - blocked edge
    - reduced capacity

### Reward Function:

Multi-objective
